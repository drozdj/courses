```python
# import libraries
import time
```

```python
# CPU (via numpy)
import numpy as np
def cpu_matrix_multiply(a, b):
    return np.matmul(a, b)
```

```python
# GPU (via PyTorch)
import torch
def gpu_matrix_multiply(a, b):
    a_tensor = torch.from_numpy(a).cuda()
    b_tensor = torch.from_numpy(b).cuda()
    return torch.matmul(a_tensor, b_tensor).cpu().numpy()
```

```python
# Generate random matrices
n = 2048*4
a = np.random.rand(n, n)
b = np.random.rand(n, n)
```

```python
# Run CPU version
start_time = time.time()
cpu_result = cpu_matrix_multiply(a, b)
cpu_time = time.time() - start_time
print(f"CPU time: {cpu_time:.6f} seconds")
```

```python
# Run GPU version
start_time = time.time()
gpu_result = gpu_matrix_multiply(a, b)
gpu_time = time.time() - start_time
print(f"GPU time: {gpu_time:.6f} seconds")
```

```python
# how much faster is the GPU?
print(f'{(cpu_time/gpu_time)*100:.2f}% difference in speed')
```
